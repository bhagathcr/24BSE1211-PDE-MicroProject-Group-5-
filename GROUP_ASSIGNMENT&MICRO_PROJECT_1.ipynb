# Quantum Confinement: Computational Analysis

## Phase 2 – Computational Focus

### Objective

The objective of this computational phase is to utilize the analytical
solution obtained in Phase 1 for a particle confined in a one-dimensional
infinite potential well.

The normalized stationary-state wave function is

Ψₙ(x,t) = √(2/L) sin(nπx/L) exp(-iEₙt/ℏ)

where

Eₙ = n²π²ℏ²/(2mL²).

The probability density is

|Ψₙ(x,t)|² = (2/L) sin²(nπx/L).

In this phase, Python, NumPy and Matplotlib are used to calculate and
visualize the standing-wave behavior and probability density for the
first three energy levels, n = 1, 2 and 3.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Computational Approach

The analytical expressions obtained in Phase 1 are implemented
numerically using Python.

The computational procedure consists of the following steps:

1. Define the physical constants and parameters of the infinite well.
2. Generate spatial points within the well.
3. Calculate the quantized energy levels for n = 1, 2 and 3.
4. Calculate the corresponding normalized wave functions.
5. Calculate the probability density |Ψ(x,t)|².
6. Plot the standing-wave patterns.
7. Plot the probability-density distributions.
8. Verify the normalization condition numerically.
9. Analyze the nodes and energy-level behavior.

```python
# Physical constants

hbar = 1.054571817e-34   # Reduced Planck's constant (J.s)
m = 9.1093837e-31        # Electron mass (kg)

# Length of the infinite potential well
L = 1e-9                 # 1 nm

# First three energy levels
n_levels = np.array([1, 2, 3])

print("Physical Parameters")
print("-------------------")
print(f"Electron mass = {m:.4e} kg")
print(f"Well length   = {L:.4e} m")
print(f"Well length   = {L * 1e9:.2f} nm")
```

## Spatial Coordinate

The electron is confined to the region

0 < x < L.

The well length is chosen as 1 nm to represent a nanoscale
confinement region. A set of equally spaced spatial points is generated
to numerically evaluate the wave functions and probability densities.

```python
# Generate spatial points inside the infinite potential well

x = np.linspace(0, L, 1000)

# Convert position to nanometres for easier visualization
x_nm = x * 1e9

print("Number of spatial points:", len(x))
print(f"Position range: {x_nm[0]:.1f} nm to {x_nm[-1]:.1f} nm")
```

## Quantized Energy Levels

From the analytical solution obtained in Phase 1, the allowed energy
states are

Eₙ = n²π²ℏ²/(2mL²),

where n = 1, 2, 3, ... is the quantum number.

The first three energy levels are calculated computationally for
n = 1, 2 and 3.

```python
# Function to calculate the quantized energy

def energy_level(n, hbar, mass, length):
    return (n**2 * np.pi**2 * hbar**2) / (2 * mass * length**2)


# Calculate energies for n = 1, 2, 3
energies_J = np.array([
    energy_level(n, hbar, m, L)
    for n in n_levels
])

# Convert joules to electron-volts
joule_to_eV = 1 / 1.602176634e-19
energies_eV = energies_J * joule_to_eV


# Display the energy values
print("Quantized Energy Levels")
print("-----------------------")

for n, E_J, E_eV in zip(n_levels, energies_J, energies_eV):
    print(f"n = {n}: {E_J:.4e} J = {E_eV:.4f} eV")
```

## Normalized Wave Function

The normalized spatial eigenfunction obtained in Phase 1 is

ψₙ(x) = √(2/L) sin(nπx/L).

The complete stationary-state wave function is

Ψₙ(x,t) = ψₙ(x) exp(-iEₙt/ℏ).

For the standing-wave visualization, the spatial part ψₙ(x) is used.
The time-dependent factor changes the phase but does not change the
probability density.

```python
# Function for the normalized spatial wave function

def wave_function(x, n, length):
    return np.sqrt(2 / length) * np.sin(
        n * np.pi * x / length
    )
```

```python
# Calculate wave functions for n = 1, 2 and 3

psi = {}

for n in n_levels:
    psi[n] = wave_function(x, n, L)

print("Wave functions calculated for n = 1, 2 and 3.")
```

## Standing-Wave Behavior

The wave functions for n = 1, 2 and 3 are plotted to visualize the
standing-wave patterns inside the infinite potential well.

As the quantum number increases, the number of wave segments increases.
The n-th energy state contains n half-wavelengths within the well.

The wave function remains zero at x = 0 and x = L, satisfying the
boundary conditions of the infinite potential well.

```python
# Plot the first three standing waves

plt.figure(figsize=(10, 6))

for n in n_levels:
    # Normalize amplitude only for visual comparison
    normalized_wave = psi[n] / np.max(np.abs(psi[n]))

    plt.plot(
        x_nm,
        normalized_wave,
        label=f"n = {n}"
    )

plt.axhline(0, linewidth=0.8)

plt.xlabel("Position x (nm)")
plt.ylabel("Normalized Wave Function")
plt.title("Standing Waves for n = 1, 2 and 3")
plt.legend()
plt.grid(True)

plt.show()
```

## Probability Density

The probability density of finding the particle at position x is given
by

|Ψ(x,t)|² = Ψ*(x,t)Ψ(x,t).

For a stationary state,

|Ψₙ(x,t)|² = (2/L) sin²(nπx/L).

The time-dependent phase factor has unit magnitude, so the probability
density does not change with time for an individual stationary state.

The probability-density distribution is therefore calculated for the
first three energy levels.


```python
# Function to calculate probability density

def probability_density(x, n, length):
    psi_n = wave_function(x, n, length)
    return np.abs(psi_n)**2


# Calculate probability densities
probability = {}

for n in n_levels:
    probability[n] = probability_density(x, n, L)

print("Probability densities calculated successfully.")
```

```python
# Plot probability densities together

plt.figure(figsize=(10, 6))

for n in n_levels:

    # Convert probability density from m^-1 to nm^-1
    probability_nm = probability[n] * 1e-9

    plt.plot(
        x_nm,
        probability_nm,
        label=f"n = {n}"
    )

plt.xlabel("Position x (nm)")
plt.ylabel(r"$|\Psi(x,t)|^2$ (nm$^{-1}$)")
plt.title("Probability Density for n = 1, 2 and 3")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Plot probability density separately for each energy level

for n in n_levels:

    plt.figure(figsize=(8, 5))

    probability_nm = probability[n] * 1e-9

    plt.plot(
        x_nm,
        probability_nm,
        linewidth=2
    )

    plt.xlabel("Position x (nm)")
    plt.ylabel(r"$|\Psi(x,t)|^2$ (nm$^{-1}$)")
    plt.title(f"Probability Density for n = {n}")
    plt.grid(True)

    plt.show()
```

## Numerical Verification of Normalization

For a normalized wave function, the total probability of finding the
particle somewhere inside the well must be equal to one:

∫₀ᴸ |Ψₙ(x,t)|² dx = 1.

Numerical integration is performed using the calculated probability
density to verify the normalization condition for n = 1, 2 and 3.

```python
# Verify normalization numerically

print("Normalization Check")
print("-------------------")

for n in n_levels:

    normalization = np.trapezoid(
        probability[n],
        x
    )

    print(
        f"n = {n}: Integral |Ψ|² dx = "
        f"{normalization:.6f}"
    )
```

## Node Analysis

A node is a position where the wave function becomes zero.

For the n-th energy state, the number of internal nodes is

Number of internal nodes = n - 1.

Therefore:

- n = 1 → 0 internal nodes
- n = 2 → 1 internal node
- n = 3 → 2 internal nodes

This demonstrates that higher-energy states have increasingly complex
standing-wave patterns.

```python
# Display the number of internal nodes

print("Internal Node Analysis")
print("----------------------")

for n in n_levels:

    nodes = n - 1

    print(
        f"n = {n}: "
        f"{nodes} internal node(s)"
    )
```

```python
# Plot the first three quantized energy levels

plt.figure(figsize=(8, 6))

for n, energy in zip(n_levels, energies_eV):

    plt.hlines(
        energy,
        0,
        1,
        linewidth=3
    )

    plt.text(
        1.02,
        energy,
        f"n = {n}, E = {energy:.3f} eV",
        verticalalignment="center"
    )

plt.xlim(0, 1.5)
plt.xticks([])

plt.ylabel("Energy (eV)")
plt.title("Quantized Energy Levels")
plt.grid(axis="y")

plt.show()
```

```python
# Compare the energy levels

print("Energy-Level Analysis")
print("=====================")

for n, energy in zip(n_levels, energies_eV):

    print(f"n = {n}")
    print(f"Energy = {energy:.4f} eV")
    print(f"Internal nodes = {n - 1}")
    print()


# Energy ratios
print("Energy Ratios")
print("-------------")

print(f"E2 / E1 = {energies_eV[1] / energies_eV[0]:.2f}")
print(f"E3 / E1 = {energies_eV[2] / energies_eV[0]:.2f}")
```

## Physical Interpretation of Results

The computational results demonstrate the quantized nature of the
energy states of an electron confined inside a one-dimensional infinite
potential well.

For n = 1, the wave function has no internal node and the probability
density has a single maximum within the well.

For n = 2, the wave function has one internal node and the probability
density has two regions of high probability.

For n = 3, the wave function has two internal nodes and the probability
density has three regions of high probability.

As the quantum number increases, the number of nodes increases and the
standing-wave pattern becomes more complex.

The energy increases according to Eₙ ∝ n². Therefore, the energy of the
second state is four times the ground-state energy, while the third state
is nine times the ground-state energy.

The probability density is zero at the boundaries of the well, which
agrees with the physical condition that the particle cannot exist
outside an infinite potential barrier.
